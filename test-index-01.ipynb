```magma

Z:=Integers();

function WeilPolynomial(L)
    P:=Parent(L);
    x:=P.1;
    g:=Z!(Degree(L)/2);
    f:=P!(x^(2*g)*Evaluate(L,1/x));
    return f;
end function;

function Getgqp(f)
    bool,p,r:=IsPrimePower(ConstantCoefficient(f));
    g:=Z!(Degree(f)/2);
    s:=Z!(r/g);
    q:=p^s;
    return g,q,p;
end function;

function WeilBaseChange(f,r)
    P:=Parent(f);
    x:=P.1;
    P1<y,u>:=PolynomialRing(Z,2);
    ff:=Evaluate(f,u);
    f2:=Resultant(u^r-y,ff,u);
    f2:=P!Evaluate(f2,[x,0]);
    return f2;
end function;
```

```magma
facs:=Factorization(100);

function simple_str(fac) 
    a:=fac[1];
    e:=fac[2];
    if e eq 1 then 
        return Sprintf("%o",a); 
    else 
        return Sprintf("%o^%o",a,e);
    end if;
end function;

function pretty_factorization(n)
    if n eq 1 then
        return "1";
    end if;

    facs:=Factorization(n);
    num_facs:=#facs;

    if num_facs gt 1 then 
        str_facs:=[simple_str(fac) : fac in facs];
        output:=&cat([str_facs[i] cat "\\cdot " : i in [1..num_facs-1]]);
        output:=output cat str_facs[num_facs];
    else 
        output:=simple_str(facs[1]);
    end if;
    
    return output;
end function;

pretty_factorization(100);
```
