```python
def level(v,Pi):
    i=0
    while not v in Pi[i]:
        i=i+1
    return i
    

def lower_neighbors(v,Pi):
    l=level(v,Pi)
    output=[]
    for w in G.neighbors_out(v):
        if level(w,Pi)==l+1:
            output.append(w)
    return output

def upper_neighbors(v,Pi):
    l=level(v,Pi)
    output=[]
    for w in G.neighbors_in(v):
        if level(w,Pi)==l-1:
            output.append(w)
    return output

def populate_below(v,sector_dict0):
    sector_dict=sector_dict0
    sector=sector_dict[v]
    X=lower_neighbors(v,Pi)
    b=sector[1]
    a=sector[0]
    n=len(X)
    h=(b-a)/n

    x0=X[0]
    sorted_neighbors=[]
    while len(X) > 1:
        sorted_neighbors.append(x0)
        X.remove(x0)
        x0=sorted(X,key=lambda x: G.distance(x,x0))[0]
    x0=X[0]
    sorted_neighbors.append(x0)
    
    for i in range(n):
        interval=[a+i*h,a+i*h+h]
        sector_dict[sorted_neighbors[i]]=interval
    return sector_dict
```
